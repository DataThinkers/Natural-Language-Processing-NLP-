```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
doc = [
    "this laptop is good",
    "this laptop is not good" 
]
```

```python
cvec=CountVectorizer(ngram_range=(1,2))
```

```python
X = cvec.fit_transform(doc)
```

```python
X.toarray()
```

```python
cvec.vocabulary_
```

```python
df = pd.DataFrame(X.toarray(),columns =cvec.get_feature_names_out())
```

```python
cvec.get_feature_names_out()
```

```python
df
```

```python
cosine_similarity(X[0],X[1])[0][0]
```
