```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import pandas as pd
```

```python
job_description = """We are looking for a Python developer with experience in Flask,
APIs, and working with databases like PostgreSQL. Familiarity with Docker is a plus."""
```

```python
resumes = [
    "Experienced Python developer. Worked on web apps using Flask and Django. Knowledge of PostgreSQL.",
    "Java developer with expertise in Spring Boot. Hands-on with MySQL and REST APIs.",
    "Python intern with experience in data analysis using Pandas and NumPy. Some exposure to Docker.",
    "Full-stack developer with skills in React and Node.js. Also worked with MongoDB.",
    "Backend developer with Flask, PostgreSQL, and deployment on Docker."
]
```

```python
documents = [job_description] + resumes
```

```python
vec = TfidfVectorizer()
```

```python
tfidf_matrix = vec.fit_transform(documents)
```

```python
sim = tfidf_matrix.toarray()
```

```python
sim[0:1]
```

```python
sim[1:]
```

```python
x= cosine_similarity(sim[0:1],sim[1:]).flatten()
```

```python
x
```

```python
df = pd.DataFrame({'Resume': resumes,'Similarity_score':x}).sort_values(by='Similarity_score',ascending=False)
```
