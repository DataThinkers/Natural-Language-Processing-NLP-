```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
import pandas as pd
```

```python
documents = [

    "I love NLP",
    "NLP is fun",
    "I love AI"
]
```

```python
 cvec = CountVectorizer(token_pattern='(?u)\\b\\w+\\b')
```

```python
bow_matrix = cvec.fit_transform(documents)
```

```python
bow_matrix
```

```python
bow_array = bow_matrix.toarray()
```

```python
bow_array
```

```python
pd.DataFrame(bow_array,columns= cvec.get_feature_names_out())
```

```python
cvec.get_feature_names_out()
```

```python
X_new = cvec.transform(["NLP is amazing"])
```

```python
X_new.toarray()
```

```python
pd.DataFrame(X_new.toarray(), columns=cvec.get_feature_names_out())
```
