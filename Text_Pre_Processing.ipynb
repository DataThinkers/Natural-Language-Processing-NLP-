### Step 10. TF-IDF 

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer(stop_words=["book","review","httpbuybookcom"], min_df=0.1)

dtm = tv.fit_transform(data["clean_text"])

df1 = dtm.toarray()

df2 = pd.DataFrame(df1, columns=cv.get_feature_names_out())
```

```python
df2
```

```python
top10_tv = df2.sum().sort_values().tail(10)
```

```python
top10_tv.plot(kind="barh")
```

```python
import matplotlib.pyplot as plt

# Create subplots: 1 row, 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# First subplot - CountVectorizer
top10_cv.plot(kind="barh", ax=ax1, color="skyblue")
ax1.set_title("Count Vectorizer")
ax1.set_xlabel("Frequency")
ax1.set_ylabel("Terms")

# Second subplot - Tfidf
top10_tv.plot(kind="barh", ax=ax2, color="salmon")
ax2.set_title("TF-IDF")
ax2.set_xlabel("Score")
ax2.set_ylabel("Terms")

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")

def text_clearning(text):
    out = text.str.lower()
    out = out.str.replace(r"[^a-z\s]","",regex=True)
    out = out.str.replace(r"\s+"," ",regex=True).str.strip()
    return out
    
def token_stopword_lemma(text):
    doc = nlp(text)
    output = [token.lemma_ for token in doc if not token.is_stop]
    return " ".join(output)

def clearning_and_preprocessing(text):
    output = text_clearning(text)
    output = output.apply(token_stopword_lemma)
    return output
```

```python
data = pd.read_csv("book_reviews_500.csv")
```

```python
data.head()
```

```python
data["clean_text"] = clearning_and_preprocessing(data["ReviewText"])
```
