```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
doc = [

    "python code python",
    "python code simple",
    "algorithms solve problems"
    
]
```

```python
vec = TfidfVectorizer()
```

```python
tfidf_matrix = vec.fit_transform(doc)
```

```python
voc =vec.get_feature_names_out()
```

```python
tfidf_array = tfidf_matrix.toarray()
```

```python
tfidf_array
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(tfidf_array, columns = voc )
print(df.round(3))
```
